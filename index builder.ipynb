```python
"""
LongEval 2025 - Month 3 INDEX BUILDING ONLY
Train: 2022-08 → Test: 2023-05
"""

import pyterrier as pt
import pandas as pd
import os
import json
from tqdm import tqdm
import shutil
import glob
import re

print("=" * 80)
print("LONGEVAL 2025 - MONTH 3 INDEX BUILDING")
print("=" * 80)

# Initialize PyTerrier
if not pt.java.started():
    pt.java.init()

# ============================================================================
# CONFIGURATION
# ============================================================================

TRAIN_BASE = "/home/cs_lobiu001/Longeval_2025_Train_Collection_p1/release_2025_p1/French/LongEval Train Collection"
TRAIN_MONTH = "2022-08"
TRAIN_DOCS = f"{TRAIN_BASE}/Trec/{TRAIN_MONTH}_fr/collection"

TEST_BASE = "/home/cs_lobiu001/LongEval_Web_2025_Test_Collection/LongEval Test Collection"
TEST_MONTH = "2023-05"
TEST_DOCS = f"{TEST_BASE}/Json/{TEST_MONTH}_fr/collection"

TRAIN_INDEX = "./train_month3_index"
TEST_INDEX = "./test_month3_index"

print(f"Configuration:")
print(f"  Train Month: {TRAIN_MONTH}")
print(f"  Test Month: {TEST_MONTH}")
print(f"  Train Docs: {TRAIN_DOCS}")
print(f"  Test Docs: {TEST_DOCS}")
print(f"  Train Index: {TRAIN_INDEX}")
print(f"  Test Index: {TEST_INDEX}")

# ============================================================================
# INDEX TRAINING DOCUMENTS
# ============================================================================

print("\n" + "=" * 80)
print("INDEXING TRAINING DOCUMENTS")
print("=" * 80)

# Check if valid index exists
train_index_valid = False
if os.path.exists(TRAIN_INDEX):
    try:
        train_index = pt.IndexFactory.of(TRAIN_INDEX)
        print(f"✓ Train index already exists: {train_index.getCollectionStatistics().getNumberOfDocuments()} docs")
        train_index_valid = True
    except:
        print("⚠️ Train index corrupted, deleting and rebuilding...")
        shutil.rmtree(TRAIN_INDEX)

if not train_index_valid:
    # Find .trec files
    files = sorted(glob.glob(os.path.join(TRAIN_DOCS, "*.trec")))
    print(f"Found {len(files)} TREC files in {TRAIN_DOCS}")
    
    if len(files) == 0:
        raise FileNotFoundError(f"No .trec files found in {TRAIN_DOCS}")
    
    print(f"Indexing {len(files)} TREC files...")
    indexer = pt.TRECCollectionIndexer(TRAIN_INDEX, verbose=True, blocks=True)
    indexer.setProperty("metaindex.compressed.reverse.allow.duplicates", "true")
    train_indexref = indexer.index(files)
    train_index = pt.IndexFactory.of(train_indexref)
    
    print(f"✓ Train index built: {train_index.getCollectionStatistics().getNumberOfDocuments()} documents")

# ============================================================================
# INDEX TEST DOCUMENTS
# ============================================================================

print("\n" + "=" * 80)
print("INDEXING TEST DOCUMENTS")
print("=" * 80)

# Check if valid index exists
test_index_valid = False
if os.path.exists(TEST_INDEX):
    try:
        test_index = pt.IndexFactory.of(TEST_INDEX)
        print(f"✓ Test index already exists: {test_index.getCollectionStatistics().getNumberOfDocuments()} docs")
        test_index_valid = True
    except:
        print("⚠️ Test index corrupted, deleting and rebuilding...")
        shutil.rmtree(TEST_INDEX)

if not test_index_valid:
    # Test data is in JSON format
    files = sorted([f for f in os.listdir(TEST_DOCS) if f.endswith(('.json', '.jsonl'))])
    print(f"Found {len(files)} JSON files in {TEST_DOCS}")
    
    if len(files) == 0:
        raise FileNotFoundError(f"No JSON files found in {TEST_DOCS}")
    
    print(f"Processing {len(files)} JSON files...")
    docs = []
    for filename in tqdm(files, desc="Loading JSON docs"):
        with open(os.path.join(TEST_DOCS, filename), 'r', encoding='utf-8') as f:
            try:
                if filename.endswith('.jsonl'):
                    for line in f:
                        if line.strip():
                            doc = json.loads(line)
                            doc_id = str(doc.get('id') or doc.get('docid') or doc.get('_id'))
                            if doc_id.startswith('doc'):
                                doc_id = doc_id[3:]
                            text = doc.get('contents') or doc.get('text') or doc.get('content', '')
                            if text:
                                docs.append({'docno': doc_id, 'text': text})
                else:
                    data = json.load(f)
                    if isinstance(data, list):
                        for doc in data:
                            doc_id = str(doc.get('id') or doc.get('docid') or doc.get('_id'))
                            if doc_id.startswith('doc'):
                                doc_id = doc_id[3:]
                            text = doc.get('contents') or doc.get('text') or doc.get('content', '')
                            if text:
                                docs.append({'docno': doc_id, 'text': text})
            except:
                pass
    
    print(f"Loaded {len(docs)} documents")
    
    print("Building test index...")
    test_indexer = pt.IterDictIndexer(TEST_INDEX, verbose=True, blocks=True)
    test_indexref = test_indexer.index(docs)
    test_index = pt.IndexFactory.of(test_indexref)
    
    print(f"✓ Test index built: {test_index.getCollectionStatistics().getNumberOfDocuments()} documents")

# ============================================================================
# SUMMARY
# ============================================================================

print("\n" + "=" * 80)
print("🎉 MONTH 3 INDEXES READY")
print("=" * 80)

print(f"\n✓ Train index: {TRAIN_INDEX}")
print(f"  Documents: {train_index.getCollectionStatistics().getNumberOfDocuments()}")

print(f"\n✓ Test index: {TEST_INDEX}")
print(f"  Documents: {test_index.getCollectionStatistics().getNumberOfDocuments()}")

print("\n✅ INDEXES SAVED! Ready for retrieval pipeline.")
```
